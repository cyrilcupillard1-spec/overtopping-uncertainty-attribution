```python
import requests
import pandas as pd
```

```python
url = "https://hubeau.eaufrance.fr/api/v2/hydrometrie/referentiel/sites"
params = {"bbox": "1.44,48.12,3.56,49.24", "size": 1000}
r = requests.get(url, params=params)
print(r.status_code)
```

```python
data=r.json()
print(data["count"])
sites=pd.DataFrame(data["data"])
sites.shape

```
