```python
import requests
import pandas as pd
```

```python
url = "https://hubeau.eaufrance.fr/api/v2/hydrometrie/referentiel/sites"
params = {"bbox": "1.44,48.12,3.56,49.24", "size": 1000}
r = requests.get(url, params=params)
print(r.status_code)
```

```python
data=r.json()
print(data["count"])
sites=pd.DataFrame(data["data"])
sites.shape
```

```python
print(sites.columns.tolist())
sites.head()
print(r.status_code, data["count"], len(sites))
```

```python
print(sites['grandeur_hydro'].value_counts())
print(sites['type_site'].value_counts())
masque=sites['grandeur_hydro']=='Q'
len(sites[masque])
sites_ok=sites[sites['type_site']=='STANDARD']




    
```

```python
print(sites_ok['surface_bv'].describe())
print(sites_ok['influence_generale_site'].value_counts())

sites_ok_filtres1=sites_ok[sites_ok['surface_bv']>=150]
sites_ok_filtres2=sites_ok_filtres1[sites_ok_filtres1['surface_bv']<=1000]
len(sites_ok_filtres2)

print(sites_ok_filtres2[['influence_generale_site', 'commentaire_influence_generale_site']].value_counts())
```
